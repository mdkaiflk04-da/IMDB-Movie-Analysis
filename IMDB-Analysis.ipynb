IMDB Movies Dataset — Visualisation Questions

```python
import matplotlib.pyplot as plt
```

Q1. Plot the average movie rating over the years (Line Chart)

```python
mry = df.groupby("year")["rating"].mean()
```

```python
plt.figure(figsize=(10,6))
plt.plot(mry.index , mry.values , color = "pink" , marker = "*" , linestyle = "--" , markersize = 9)
plt.xlabel("year")
plt.ylabel("Averagw Rating")
plt.grid(alpha = 0.8)
plt.show()
```

Q2. Plot the number of movies released per year (Line Chart)

```python
mpy = df.groupby("year")["id"].count()
```

```python
plt.plot(mpy.index , mpy.values)
plt.show()
```

Q3. Plot the distribution of movie ratings (Histogram using Matplotlib)

```python
plt.figure(figsize = (10,6))
plt.hist(df["rating"] , color = "magenta" , edgecolor = "black" , bins =25 )
plt.xlabel("Rating")
plt.ylabel("Frequency")
plt.title("Distribution of Movie Ratings")
plt.grid(alpha = 0.3)
plt.show()
```

Q4. Plot the distribution of movie ratings (Histogram using Seaborn)

```python
import seaborn as sns

plt.title("Distribution of Movie Ratings")
sns.histplot(df["rating"], bins = 20 , edgecolor = "black")
plt.show()
```

Q5. Which are the top 10 most frequent actors? (Horizontal Bar Chart)

```python
fa = df["stars_name"].str.split(",").explode().value_counts().head(10)

```

```python
plt.figure(figsize=(10,6))
plt.bar(fa.index, fa.values , color = ["blue","green","yellow"], width = 0.5)
plt.xticks(rotation = 45)
plt.show()
```

Q6. Plot the relationship between year and rating (Scatter Plot)

```python
plt.scatter(df["year"] , df["rating"] , color = "magenta" , edgecolor = "black")
plt.title("Relationship between Year and Rating")
plt.xlabel("Year")
plt.ylabel("Rating")
plt.show()
```
